## 2.  Проверка ошибок в данных и их предобработка


### 2.1. Названия, или метки, столбцов датафрейма


```python
df.columns
```

```python
# Переводим названия столбцов в стиль snake case
df.columns = df.columns.str.lower().str.replace(' ','_')
```

```python
df.columns
```

### 2.2. Типы данных


```python
# переведем год выпуска в числовой формат int64 т.к. у нас есть только год, без конкретной даты
df['year_of_release'] = df['year_of_release'].astype('Int64')
df['year_of_release']
```

```python
# переведем значения в полях с продажами и оценкой пользователей в float
df[['eu_sales', 'jp_sales','user_score']] = df[['eu_sales', 'jp_sales','user_score']].apply(pd.to_numeric, errors = 'coerce')
df.info()
```

мы перевели значения в float, заменив значения unkown на NaN

- В числовых столбцах могут встретиться строковые значения, например `unknown` или другие. Приводите такие столбцы к числовому типу данных, заменив строковые значения на пропуски.

### 2.3. Наличие пропусков в данных


```python
# кол-во пропусков
df.isna().sum()
```

```python
## относительное кол-во пропусков
df.isna().sum() / len(df) * 100
```

- Пропуски встречаются в name, year of release, genre, critic score, user score, rating
- Больше всего пропуски в critic score (8714 - 51,4%), user score и rating (около 6800 - 40%) что является значительным показателем
- Данные в этих столбцах могут отсутствовать из-за того что ESRB рейтинг был принят в 1994 году, и некоторые игры были выпущены до этого и не получили свой рейтинг, то же самое может быть и с оценками
- можно удалить строки где есть пропуски у столбцов name, year of release, genre так как число пропусков не превышает 2%
- для столбцов с оценками и рейтинга пропуски можно заменить на индикатор или "нет оценки", мы не можем удалить эти значения (они составляют половину данных)

```python
# Удаляем строки, где пропущены данные
df = df.dropna(subset=['name', 'genre', 'year_of_release'])
```

```python
df.isna().sum()
```

```python
# Заполняем пропуски в продажах средним значением для той же платформы и года
for col in ['eu_sales', 'jp_sales']:
    df[col] = df[col].fillna(df.groupby(['platform', 'year_of_release'])[col].transform('mean'))
    df[col] = df[col].fillna(0)
    # Если после этого остались пропуски (например, игра единственная на платформе за этот год), 
# просто заполняем их нулями


```

```python
df.isna().sum()
```

```python
# заменим пропуски в рейтинге на unknown
df['rating'] = df['rating'].fillna('unknown')
```

```python
df[['user_score','critic_score']] = df[['user_score','critic_score']].fillna(-1)
```

### 2.4. Явные и неявные дубликаты в данных

```python
# сделаем названия игр с заглавной буквы
df['name']= df['name'].str.title()
```

```python
df.groupby('genre')['genre'].count()
```

```python
# находим количество платформ
df.groupby('platform')['platform'].count()
```

```python
# ищем количество рейтингов
df.groupby('rating')['rating'].count()
```

```python
# заменим рейтинг К-А на Е, так как К-А устарел
df['rating'] = df['rating'].str.replace('K-A', 'E')
```

```python
# ищем количество игр по годам
df.groupby('year_of_release')['year_of_release'].count()
```

```python
# выявляем количество явных дубликатов в датафрейме
df.duplicated().sum()
```

```python
# количество строк ДО удаления
rows_before = len(df)

# удаляем дубликаты и сбрасываем индексы
df.drop_duplicates(inplace=True)
df = df.reset_index(drop=True)

# сохраняем количество строк ПОСЛЕ удаления
rows_after = len(df)

# считаем разницу
absolute_deleted = rows_before - rows_after
relative_deleted = (absolute_deleted / rows_before) * 100

# выводим результат
print(f"Удалено явных дубликатов (абсолютное значение): {absolute_deleted}")
print(f"Процент удаленных дубликатов (относительное значение): {relative_deleted:.2f}%")
```
